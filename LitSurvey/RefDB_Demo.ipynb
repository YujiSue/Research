```python
#@title Install required modules

import json
import os
import pandas as pd
from pprint import pprint
import textwrap

!pip install git+https://github.com/YujiSue/ysref.git &> /dev/null

from ysref.refdb import RefDB
from ysref.pubmed import getPMSummary,getPMList,getPMAbstract
from ysref.dlfull import getFullText
from ysref.mining import mineWord,mineWordFrom
```

# 2. Connect/Create database file

### <font color="red">\* If you have a database previously used, please upload it before running the program.</font>  

### When running the program below, if a database already exists, it will be opened and new records can be added. If no database is found, a new one will be created automatically.

```python
#@title Connect DB
#@markdown Please change the database path if you need
dbpath = '/content/reference.db' #@param {"type":"string"}
db = RefDB(dbpath)
```

# 3. PubMed search

### \*As noted in the first section, a NCBI API key is required. Please obtain your key, register it in Google Colab Secrets under the name "NCBI_API", and make sure it is set to be accessible.

```python
#@title Run
#@markdown Set query keyword
query = 'elegans' #@param {"type":"string"}
#@markdown Set period to search
start_date = '2020/04/01' #@param {"type":"string"}
end_date = '2020/04/30' #@param {"type":"string"}

# Call PubMed search function
reflist = getPMList(query,
                    key='colab',
                    condition={
                        'period': [start_date, end_date]
                        },
                    verbose=True)

# Display results
refDF = pd.DataFrame(reflist)
display(refDF.head(20).style.hide(axis="index"))
refDF.to_csv('/refsample.tsv', sep="\t")
```

# 4. Store results to the database

```python
#@title Insert records into DB
#@markdown Set label
label = '2020_Apr' #@param {"type":"string"}

db.registerRefDB(reflist, label)
```

```python
#@title (Optional) Check reference information
#@markdown Set PubMed ID to check
pubmed_id = '32269224' #@param {"type":"string"}

pprint(db.getRefSummary(pubmed_id))
```

```python
#@title (Optional) Display article abstract.
#@markdown Set PubMed ID to show
pubmed_id = '32269224' #@param {"type":"string"}

# Call function
abst = getPMAbstract(pubmed_id)

# IF successed
if abst:
  print('Abstract:')
  print(textwrap.fill(abst, width=60))
# IF failed
else:
  print("Failed to get abstract from PubMed.")
```

```python
#@title (Optional) Export reference list as bibtex format (.bib)
#@markdown Reference list will be exported to '/content/reference.bib'.<br/>
#@markdown You can use the list to import information of found articles into reference management software such as Zotero, Mendeley, Papers, and EndNote.

db.export(f'/content/reference.bib')
```

```python
#@title (Demo) Download full-text article
!mkdir -p fulltext

pubmed_id = '32269224' #@param {"type":"string"}
refinfo = db.getRefSummary(pubmed_id)
result = getFullText(pubmed_id, refinfo['doi'], '/content/fulltext', verbose=True)
pprint(result)
```

## Bulk processing and database recording

```python
#@title Run
!mkdir -p fulltext
# Get reference IDs to full-text downlaod. If 'file' is already registered, ignore the record.
refids = db.getRefIDs(conditions=f"file=''")
# Try downloading for each article
for refid in refids:
  ## Get DOI
  refinfo = db.getRefSummary(refid)
  ## Try to download full-text
  result = getFullText(refid, refinfo['doi'], '/content/fulltext')
  ## IF successed
  if result['status']:
    ### Update URL, file and logs
    db.updateRecord(refid, ['link','file','attribute'], [
        result['fulltexts'],
        os.path.split(result['path'])[1],
        {k:v for k,v in result.items() if k not in ['status', 'msg', 'path', 'fulltexts']}
    ])
  ## IF failed
  else:
    ### Save the error log
    db.updateRecord(refid, ['link','attribute'], [
        result['fulltexts'],
        {'error':result['msg']}
    ])
```

# 6. Text mining

```python
#@title (Demo) Mining keywords
from pprint import pprint
result = {}
#@markdown Set PubMed ID
pubmed_id = '32269224' #@param {"type":"string"}
#@markdown Set keywords in regex format
keywords = 'tm[1-9]+[0-9]{2,4}' #@param {"type":"string"}
mineWordFrom(result, keywords, f'/content/fulltext/{pubmed_id}', verbose=True)
pprint(result)
os.chdir('/content')
```

## Bulk processing and database recording

```python
#@title Run
#@markdown Set keywords
keywords = 'tm[1-9]+[0-9]{2,4}' #@param {"type":"string"}

# Get reference IDs to analyze. If 'file' is NOT registered, ignore the record.
refids = db.getRefIDs(conditions=f"file!=''")
# Try mining for each article
for refid in refids:
  ## Init. container
  result = {}
  ## Call recursive function
  mineWordFrom(result, keywords, f'/content/fulltext/{refid}')
  if 0 < len(result):
    note = f"Text mining: {','.join(result.keys())}"
    db.updateRecord(refid, ['note'], [note])
#
os.chdir('/content')
```
